```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv('./drive/MyDrive/yt.csv')
data=data.drop_duplicates()
data
```

```python
data.info()
```

```python
print(f'There are {data.shape[0]} rows')
print(f'There are {data.shape[1]} columns')
```

```python
data["Subscribers"]=pd.to_numeric(data["Subscribers"], errors="coerce")
data["Subscribers"]=data["Subscribers"].fillna(0).astype("int64")

data1=data.sort_values(by="Subscribers",ascending=False).head(10)
print("These channels have the most subscribers:")
data1.loc[:, ["Rank", "Channel name", "Subscribers"]]
```

```python
data["Video Uploads"]=pd.to_numeric(data["Video Uploads"], errors="coerce")
data["Video Uploads"]=data["Video Uploads"].fillna(0).astype("int64")

data1=data.sort_values(by="Video Uploads",ascending=False).head(10)

print("These channels have the most uploaded videos:")
data1.loc[:, ["Rank", "Channel name", "Video Uploads"]]
```

```python
data1=data.sort_values(by="Video views",ascending=False).head(10)

print("These channels have the most views:")
data1.loc[:, ["Rank", "Channel name", "Video views"]]
```

```python
data2=pd.DataFrame(data[data['Channel name'].str.contains('VEVO')])
data2=data2.sort_values(by='Subscribers',ascending=False).head(10)
print(f'These music artist channels have the most subscribers: \n')
data2.loc[:,['Channel name','Subscribers','Rank']]
```

```python
data2=pd.DataFrame(data[data['Channel name'].str.contains('VEVO')])
data2=data2.sort_values(by='Video views',ascending=False).head(10)
print(f'These music artist channels have the most views: \n')
data2.loc[:,['Channel name','Video views','Rank']]
```

```python
import matplotlib.pyplot as plt
data["Subscribers"]=pd.to_numeric(data["Subscribers"], errors="coerce")
data["Subscribers"]=data["Subscribers"].fillna(0).astype("int64")

plt.scatter(data["Subscribers"], data["Video views"])
plt.xlabel("Subscribers")
plt.ylabel("Video views")
plt.title("Subscribers vs Video views")
plt.show()

cor2=data['Subscribers'].corr(data['Video views'])
print(f'\nThe correlation between the number of subscribers of the channel and the number of views : {cor2}\n')
print('Hence the correlation is moderate to high.')
```

```python
data["Subscribers"]=pd.to_numeric(data["Subscribers"], errors="coerce")
data["Subscribers"]=data["Subscribers"].fillna(0).astype("int64")
data["Video Uploads"]=pd.to_numeric(data["Video Uploads"], errors="coerce")
data["Video Uploads"]=data["Video Uploads"].fillna(0).astype("int64")

plt.scatter(data["Subscribers"], data["Video Uploads"])
plt.xlabel("Subscribers")
plt.ylabel("Video Uploads")
plt.title("Subscribers vs Video Uploads")
plt.show()

cor3=data['Subscribers'].corr(data['Video Uploads'])
print(f'\nThe correlation between the number of subscribers of the channel and the number of videos uploaded : {cor3}\n')
print('Hence the correlation is very low, or negligible.')
```

```python
data["Video Uploads"]=pd.to_numeric(data["Video Uploads"], errors="coerce")
data["Video Uploads"]=data["Video Uploads"].fillna(0).astype("int64")

plt.scatter(data["Video views"], data["Video Uploads"])
plt.xlabel("Video views")
plt.ylabel("Video Uploads")
plt.title("Video Views vs Video Uploads")
plt.show()

cor4=data['Video views'].corr(data['Video Uploads'])
print(f'\nThe correlation between the number of views the channel has and the number of videos uploaded : {cor4}\n')
print('Hence the correlation is very low, or negligible.')
```

```python
data1=data.drop('avg_views',axis=1)
data1['Average Views']=data1['Video views']/data1['Video Uploads']
data1.head(5)
```

```python
data["Video Uploads"]=pd.to_numeric(data["Video Uploads"], errors="coerce")
data["Video Uploads"]=data["Video Uploads"].fillna(0).astype("int64")
data["Subscribers"]=pd.to_numeric(data["Subscribers"], errors="coerce")
data["Subscribers"]=data["Subscribers"].fillna(0).astype("int64")
data1=data.drop('Channel name',axis=1)
data2=data1.drop('avg_views',axis=1)
data2['Grade']=data2['Grade'].map({'A++ ':5,'A+ ':4,'A ':3,'A- ':2,'B+ ':1,})
data2['Rank']=data2['Rank'].str[0:-2]
data2['Rank']=data2['Rank'].str.replace(',','')
data2.corr()
```

```python
import seaborn as sb

data["Video Uploads"]=pd.to_numeric(data["Video Uploads"], errors="coerce")
data["Video Uploads"]=data["Video Uploads"].fillna(0).astype("int64")
data["Subscribers"]=pd.to_numeric(data["Subscribers"], errors="coerce")
data["Subscribers"]=data["Subscribers"].fillna(0).astype("int64")
data1=data.drop('Channel name',axis=1)
data2=data1.drop('avg_views',axis=1)
data2['Grade']=data2['Grade'].map({'A++ ':5,'A+ ':4,'A ':3,'A- ':2,'B+ ':1,})
data2['Rank']=data2['Rank'].str[0:-2]
data2['Rank']=data2['Rank'].str.replace(',','')

sb.barplot(x='Grade',y='Subscribers',data=data2,linewidth=2)
plt.figure(figsize=(100,10))
plt.show()

sb.barplot(x='Grade',y='Average Views',data=data2)
plt.show()

sb.barplot(x='Grade',y='Video Uploads',data=data2)
plt.show()

sb.barplot(x='Grade',y='Average Views',data=data2)
plt.show()
```
